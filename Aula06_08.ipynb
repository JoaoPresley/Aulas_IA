```python
# Objetivo do 8-Puzzle
# A meta é deixar os numeros de 1 a 8 em ordem
# o zero representa o espaço vazio :)
GOAL = (1, 2, 3, 4, 5, 6, 7, 8, 0)


# Exibir em formato de tabuleiro 3x3

def show (state: tuple):
  # 1°: Inicial, 2°: final, 3°: incremento
    for i in range(0, 9, 3):
        print(state[i:i+3])

# Mapear todas as possibilidades de movimentações
# up    = mover para cima
# down  = maver para baixo
# left  = mover para esquerda
# right = mover para direita

# Pensando na tupla
# up    = p - 3
# down  = p + 3
# left  = p - 1
# right = p + 1
MOVES = {"up": -3,
         "down": +3,
         "left": -1,
         "right": +1}

# Função para encontrar o 0 (espaço vazio)
# retorna a posição do 0
def find_zero(state: tuple) -> int:
    return state.index(0)


# Função para identificar movimentações possíves
def valid_movies(z: int):
    # retorna a r (linha) e c (coluna)
    r, c = divmod(z,3)

    # moves = movimentos permitidos
    moves =[]
    if r > 0:
        moves.append("up")
    if r < 2:
        moves.append("down")

    if c > 0:
        moves.append("left")
    if c < 2:
        moves.append("right")

    return moves

# Função para aplicar movimentação
def apply_move(state: tuple, move: str):
    # Posição do zero
    z = find_zero(state)

    # Nova posição do zero
    nz = z + MOVES[move]

    # troca de valores
    s = list(state)
    s[z], s[nz] = s[nz], s[z]

    return tuple(s)

```

```python
# busca em largura -> BFS
def bfs(start: tuple):
  # Fronteiras sempre POPULADAS
  # converter o start
  # list(start)
  frontiers = deque([[start]])

  # gerenciar os visitados
  visited = {start}

  # enquanto houver fronteiras
  while frontiers:
    path = frontiers.popleft()
    # sempre armazeno o último estado do path
    state = path[-1]
    # verifico se cheguei no estado final
    if state == GOAL:
      return path

    # Encontrar a posição de 0
    z = find_zero(state)
    # valido movimentações possiveis
    for m in valid_movies(z):
      # novo estado
      new_state = apply_move(state, m)

      if new_state not in visited:
        visited.add(new_state) # atualizo visitados
        # atualizo a fronteira
        frontiers.append(path + [new_state])

  # Não encontrou solução
  return None
```

```python
# Estado inicial
start = (1, 2, 3,
         4, 5, 6,
         0, 7, 8)

print("Estado inicial: ")
# exibir estado inicial
show(start)

# Executar a busca em largura para encontrar o caminho
solution = bfs(start)

if solution is None:
  print("Não foi encontrada solução")
else:
  step = len(solution) - 1
  print(f"Solução encontrada em {step} passos. \nReproduzindo caminho encontrado: ")
  for st in solution:
    show(st)
```
